```python
import pandas as pd
import numpy as np
```

```python
pd.set_option('display.max_columns',None)
```

```python
df = pd.read_csv('mushrooms.csv')
df.head()
```

```python
df.columns
```

```python
for col in df.columns:
    print(df[col].unique())
    print(df[col].nunique())
    print("--"*50)
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
for column in df.columns:
    df[column] = le.fit_transform(df[column])
```

```python
df.head()
```

```python
X = df.drop('class', axis=1)
y = df['class']
```

```python
X
```

```python
y
```

```python
from sklearn.decomposition import PCA
pca1 = PCA(n_components=7)
pca_fit = pca1.fit_transform(X)
```

```python
pca1.explained_variance_ratio_
```

```python
sum(pca1.explained_variance_ratio_)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(pca_fit,y, test_size =0.2, random_state =42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC

from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
```

```python
lr = LogisticRegression()
lr.fit(X_train, y_train)

knn = KNeighborsClassifier()
knn.fit(X_train, y_train)

dtr = DecisionTreeClassifier()
dtr.fit(X_train, y_train)

svc = SVC()
svc.fit(X_train, y_train)

rf = RandomForestClassifier()
rf.fit(X_train,y_train)

gb = GradientBoostingClassifier()
gb.fit(X_train, y_train)

```

```python
y_predict1 = lr.predict(X_test)
y_predict2 = knn.predict(X_test)
y_predict3 = svc.predict(X_test)
y_predict4 = dtr.predict(X_test)
y_predict5 = rf.predict(X_test)
y_predict6 = gb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("Acc LR", accuracy_score(y_test, y_predict1))
print("Acc KNN", accuracy_score(y_test, y_predict2))
print("Acc SVC", accuracy_score(y_test, y_predict3))
print("Acc DTR", accuracy_score(y_test, y_predict4))
print("Acc RF", accuracy_score(y_test, y_predict5))
print("Acc GB", accuracy_score(y_test, y_predict6))
```

```python
acc_score_df = pd.DataFrame({"Models":["LR","KNN","SVC","DTR","RF","GB"],
              "Acc_score":[accuracy_score(y_test,y_predict1),accuracy_score(y_test,y_predict2),accuracy_score(y_test,y_predict3),accuracy_score(y_test,y_predict4),accuracy_score(y_test,y_predict5),accuracy_score(y_test,y_predict6)]})
```
